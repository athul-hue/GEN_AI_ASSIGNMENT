## Exploring Text Generation Using GPT-2

### Objective
The objective of this assignment is to understand how different text-generation parameters influence the output produced by a language model.

In this assignment, the pretrained GPT-2 model from the Hugging Face Transformers library is used to generate text with different parameter settings.

```python
from transformers import GPT2LMHeadModel, GPT2Tokenizer
import torch
print("Libraries imported successfully!")
```

```python
tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
model = GPT2LMHeadModel.from_pretrained("gpt2")
tokenizer.pad_token = tokenizer.eos_token
print("GPT-2 model loaded successfully!")
```

## Original Prompt
The opening sentence used for all five experiments is:

**"The future of artificial intelligence will"**

The same prompt is used in each experiment so that the effect of changing generation parameters can be observed.


```python
def generate_text(prompt, temperature=1.0, top_k=50, top_p=1.0, repetition_penalty=1.0, seed=42):
    torch.manual_seed(seed)
    inputs = tokenizer(prompt, return_tensors="pt")
    with torch.no_grad():
        output = model.generate(
            **inputs,
            max_new_tokens=60,
            temperature=temperature,
            top_k=top_k,
            top_p=top_p,
            repetition_penalty=repetition_penalty,
            do_sample=True,
            pad_token_id=tokenizer.eos_token_id
        )
    return tokenizer.decode(output[0], skip_special_tokens=True)
```

## Experiment 1 — Low Temperature
**Parameter changed:** Temperature  
**Value:** 0.2

**Observation:** More predictable and focused.


```python
prompt = "The future of artificial intelligence will"
output = generate_text(prompt, temperature=0.2, top_k=50, top_p=1.0, repetition_penalty=1.0, seed=42)
print(output)
```

## Experiment 2 — High Temperature
**Parameter changed:** Temperature  
**Value:** 1.5

**Observation:** More creative and random; may become less coherent.


```python
prompt = "The future of artificial intelligence will"
output = generate_text(prompt, temperature=1.5, top_k=50, top_p=1.0, repetition_penalty=1.0, seed=42)
print(output)
```

## Experiment 3 — Top-k
**Parameter changed:** Top-k  
**Value:** 10

**Observation:** Restricts choices to likely tokens and keeps output focused.


```python
prompt = "The future of artificial intelligence will"
output = generate_text(prompt, temperature=1.0, top_k=10, top_p=1.0, repetition_penalty=1.0, seed=42)
print(output)
```

## Experiment 4 — Top-p
**Parameter changed:** Top-p  
**Value:** 0.5

**Observation:** Restricts choices using cumulative probability.


```python
prompt = "The future of artificial intelligence will"
output = generate_text(prompt, temperature=1.0, top_k=50, top_p=0.5, repetition_penalty=1.0, seed=42)
print(output)
```

## Experiment 5 — Repetition Penalty
**Parameter changed:** Repetition Penalty  
**Value:** 1.5

**Observation:** Discourages repeated words and phrases.


```python
prompt = "The future of artificial intelligence will"
output = generate_text(prompt, temperature=1.0, top_k=50, top_p=1.0, repetition_penalty=1.5, seed=42)
print(output)
```

# Comparison of Five Experiments

| Experiment | Parameter Changed | Value | Expected Effect |
|---|---|---:|---|
| 1 | Temperature | 0.2 | More predictable and focused |
| 2 | Temperature | 1.5 | More creative and random |
| 3 | Top-k | 10 | More restricted and focused |
| 4 | Top-p | 0.5 | More controlled token selection |
| 5 | Repetition Penalty | 1.5 | Less repetition |

The observations should be checked against the actual generated outputs above.


# Custom Prompt

**"A young student discovered a mysterious device in the university laboratory."**

Three different versions will be generated using different settings.


## Version 1 — Controlled
**Temperature:** 0.5


```python
custom_prompt = "A young student discovered a mysterious device in the university laboratory."
custom_output = generate_text(custom_prompt, temperature=0.5, top_k=50, top_p=1.0, repetition_penalty=1.0, seed=105)
print(custom_output)
```

## Version 2 — Balanced
**Temperature:** 0.9


```python
custom_prompt = "A young student discovered a mysterious device in the university laboratory."
custom_output = generate_text(custom_prompt, temperature=0.9, top_k=50, top_p=1.0, repetition_penalty=1.0, seed=109)
print(custom_output)
```

## Version 3 — Creative
**Temperature:** 1.3


```python
custom_prompt = "A young student discovered a mysterious device in the university laboratory."
custom_output = generate_text(custom_prompt, temperature=1.3, top_k=50, top_p=1.0, repetition_penalty=1.0, seed=113)
print(custom_output)
```

# Custom Generation Analysis

The three versions demonstrate how temperature affects GPT-2 output. Lower temperature generally produces more controlled and predictable text. A medium temperature provides a balance between predictability and creativity. Higher temperature produces more varied and random text, although coherence may decrease.


# Overall Analysis

The experiments demonstrate that GPT-2 generation parameters influence predictability, creativity, randomness, and repetition. Lower temperature generally makes output more focused, while higher temperature increases randomness. Lower top-k and top-p values restrict possible next tokens. A higher repetition penalty discourages repeated words and phrases.


# Conclusion

This assignment demonstrated text generation using the pretrained GPT-2 model. Changing temperature, top-k, top-p, and repetition penalty produced different styles of generated text. These parameters provide control over creativity, randomness, predictability, and repetition.
